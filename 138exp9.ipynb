```python
data= data.sample(frac = 0.1,random_state=1)
data.shape
```

```python
Fraud = data[data['Class']==1]
Valid = data[data['Class']==0]
outlier_fraction = len(Fraud)/float(len(Valid))
print(outlier_fraction)
print('Fraud Cases: {}'.format(len(Fraud)))
print('Valid Cases: {}'.format(len(Valid)))
```

```python
import seaborn as sns
corrmat=data.corr()
top_corr_features=corrmat.index
plt.figure(figsize=(20,20))
g=sns.heatmap(data[top_corr_features].corr(),annot=True,cmap="RdYlGn")
#plot heat map
```

```python
columns = data.columns.tolist()
columns = [c for c in columns if c not in ["Class"]]
target = "Class"
X = data[columns]
Y = data[target]
X_outliers = np.random.uniform(low=0,high=1,size=(X.shape[0],X.shape[1]))
print(X.shape)
print(Y.shape)
```

```python
classifiers = {
    "Isolation Forest":IsolationForest(n_estimators=100,max_samples=len(X),
                                       contamination=min(outlier_fraction, 0.5),random_state=RANDOM_SEED,
                                       verbose=0),
    "Local Outlier Factor":LocalOutlierFactor(n_neighbors=20,algorithm='auto',leaf_size=30,
                                              metric='minkowski',p=2,metric_params=None,contamination=min(outlier_fraction, 0.5))}
```

```python
type(classifiers)
n_outliers=len(Fraud)
for i,(clf_name,clf) in enumerate(classifiers.items()):
  if clf_name=="Local Outlier Factor":
    y_pred = clf.fit_predict(X)
    scores_pred=clf.negative_outlier_factor_
  else:
    clf.fit(X)
    scores_pred=clf.decision_function(X)
    y_pred=clf.predict(X)
    y_pred[y_pred==1]=0
  y_pred[y_pred==-1]=1
  n_errors=(y_pred!=Y).sum()
  print('{}: {}'.format(clf_name,n_errors))
  print(accuracy_score(Y,y_pred))
  print(classification_report(Y,y_pred))
  print("Classification Report")
  print(classification_report(Y,y_pred))
```
